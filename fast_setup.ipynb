```python
# 5. Start Service with ngrok
import os
from pyngrok import ngrok
import threading
import time

# Set your ngrok token here (Get one free at https://dashboard.ngrok.com/get-started/your-authtoken)
NGROK_TOKEN = "YOUR_NGROK_TOKEN_HERE" 
ngrok.set_auth_token(NGROK_TOKEN)

def start_comfy():
 os.system('cd /kaggle/working/ComfyUI && python main.py --listen 0.0.0.0 --port 8188')

# Start ComfyUI in background
t = threading.Thread(target=start_comfy)
t.daemon = True
t.start()

# Wait for startup
time.sleep(15)

# Open ngrok tunnel
public_url = ngrok.connect(8188)
print(f"🌍 Your ComfyUI is live at: {public_url}")
```
