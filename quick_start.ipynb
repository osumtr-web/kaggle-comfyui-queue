```python
# --- 2. Start Service & ngrok Tunnel ---
import os, threading, time
from pyngrok import ngrok

# TODO: Replace with your actual ngrok token from https://dashboard.ngrok.com/get-started/your-authtoken
NGROK_TOKEN = "YOUR_NGROK_TOKEN_HERE" 
ngrok.set_auth_token(NGROK_TOKEN)

def start_comfy():
 os.system('cd /kaggle/working/ComfyUI && python main.py --listen 0.0.0.0 --port 8188')

# Start ComfyUI in background
t = threading.Thread(target=start_comfy)
t.daemon = True
t.start()

print("⏳ Starting ComfyUI... please wait ~30s")
time.sleep(30)

# Open ngrok tunnel
public_url = ngrok.connect(8188)
print(f"\n🌍 SUCCESS! Your ComfyUI is live at: {public_url}")
print("👉 Click the link above to access your GPU server.")
```
