```python
import os, time, subprocess, threading
os.chdir('/kaggle/working/ComfyUI')

# 6. Watchdog: Restart ComfyUI if it crashes
def start_comfy_watchdog():
    while True:
        print("🔄 Starting ComfyUI...")
        proc = subprocess.Popen(['python', 'main.py', '--listen', '0.0.0.0', '--port', '8188'])
        proc.wait()
        print("⚠️ ComfyUI crashed. Restarting in 5s...")
        time.sleep(5)

# Start Watchdog in background
t = threading.Thread(target=start_comfy_watchdog)
t.daemon = True
t.start()

print("⏳ Waiting 15 seconds for ComfyUI to initialize...")
time.sleep(15)

# 7. Start LocalTunnel
print("🌍 Starting LocalTunnel...")
!npx localtunnel --port 8188
```
